# Common Regex Patterns in Python

## 1. `\d+`
**Meaning:** One or more digits.

**Examples:** `2026`, `123`, `98765`

## 2. `\s+`
**Meaning:** One or more whitespace characters, such as spaces, tabs, or newlines.

**Examples:** Extra spaces, tabs, multiple line breaks.

## 3. `https?://\S+`
**Meaning:** A simple URL pattern that matches URLs beginning with `http://` or `https://`.

**Examples:** `https://example.com`, `http://google.com`

## 4. `<.*?>`
**Meaning:** A basic HTML-tag pattern that matches text starting with `<` and ending at the nearest `>`.

**Examples:** `<p>`, `</p>`, `<h1>`

## 5. `[^a-zA-Z\s]`
**Meaning:** Matches any character other than English letters (`a-z`, `A-Z`) and whitespace.

**Examples:** `123`, `!`, `@`, `#`

**Note:** This pattern matches one character at a time. Use `+` (`[^a-zA-Z\s]+`) to match one or more consecutive characters that are not English letters or whitespace.

```python
import re
import emoji
from collections import Counter
from textblob import TextBlob
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords


def Text_Cleaner(text):
    Text=text.copy()

    # Lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Remove Digits
    text = re.sub(r'\d+', '', text)

    # Remove punctuation
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Spelling correction
    text = str(TextBlob(text).correct())

    # Convert emojis to text
    text = emoji.demojize(text)

    # Stopword Removal
    stop_words = set(stopwords.words('english'))

    words = []

    for word in text.split():

        if word.upper() in chat_words:
            word = chat_words[word.upper()]

        if word.lower() not in stop_words:
            words.append(word)

    # Stemming + Lemmatization
    stemmer = PorterStemmer()
    lemmatizer = WordNetLemmatizer()

    cleaned_words = []

    for word in words:

        word = stemmer.stem(word)

        word = lemmatizer.lemmatize(word, pos='v')
        word = lemmatizer.lemmatize(word, pos='n')
        word = lemmatizer.lemmatize(word, pos='a')

        cleaned_words.append(word)

    Text = Text.split()
    Frequency = Counter(Text)



    return {
            "Text" : " ".join(cleaned_words) ,
            "Frequency : " : Frequency,
            "Vocabulary" : set(Text)

            }
```

```python
def Word_Frequency(text):
  Text_Cleaner(text)
  Text_Freq={}
  for i in text.split():
    if i not in Text:
      Text[i]=1
    else:
      Text[i]+=1
  return Text_Freq
```
