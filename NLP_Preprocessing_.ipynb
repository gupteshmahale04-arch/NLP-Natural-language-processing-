# Preprocessing of data Function




```python
import re
import string
import emoji
from collections import Counter
from textblob import TextBlob

def clean_text(text):
    text = text.lower()                                # lowercase
    text = re.sub(r"<.*?>", "", text)                  # HTML tags
    text = re.sub(r"http\S+|www\.\S+", "", text)       # URLs
    text = re.sub(r"\d+", "", text)                    # numbers
    text = text.translate(str.maketrans("", "", string.punctuation))  # punctuation
    text = re.sub(r"\s+", " ", text).strip()           # extra spaces
    text = str(TextBlob(text).correct())               # Spelling Correction
    text = emoji.demojize(text)                        # Emojis to text

    return text

clean_text(text)
# hello visit in

```

# Stemming and Lemmatization

```python

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
cleaned_words = []

for word in text:

        word = stemmer.stem(word)
        word = lemmatizer.lemmatize(word, pos='v')
        word = lemmatizer.lemmatize(word, pos='n')
        word = lemmatizer.lemmatize(word, pos='a')

        cleaned_words.append(word)
```

# Bag of Words

```python
reviews = ['food was good',
           'delivery was late',
           'food was cold and late']
vocab = []

for review in reviews:
  for word in review.lower().split():
    if word not in vocab:
      vocab.append(word)
print(vocab)
```

Build the Bag of Words matrix

```python
bow=[]
for review in reviews:
  tokens = review.lower().split()
  vector=[]
  for word in vector:
     vector.append(tokens.count(word))
     bow.append(vector)
  print(vector)
```
